#### Task 5.2 Perform the Analysis (20 points)

- Perform the proposed analysis.
- Be sure to comment your code throughout so that it is easy to follow. 

*You may add more code cells if needed.* 


</div>

```r
library(tidyverse)
library(tidymodels)

# Select the variables needed for the analysis
knn_data <- womens_march_madness_tidy |>
  select(seed, conf_wins, conf_wins_pct, tourney_finish) |>
  drop_na() |>
  mutate(tourney_finish = as_factor(tourney_finish))

# Set the random seed
set.seed(123)

# Split the data into training and testing sets
data_split <- initial_split(knn_data, prop = 0.75,
                            strata = tourney_finish)

# Create training and testing sets
training_data <- training(data_split)
testing_data <- testing(data_split)

# Scale and center the predictor variables
knn_recipe <- recipe(tourney_finish ~ seed + conf_wins +
                       conf_wins_pct,
                     data = training_data) |>
  step_scale(all_predictors()) |>
  step_center(all_predictors())

# Create a KNN classification model with 5 neighbors
knn_spec <- nearest_neighbor(weight_func = "rectangular",
                             neighbors = 5) |>
  set_engine("kknn") |>
  set_mode("classification")

# Combine the recipe and model into a workflow
knn_workflow <- workflow() |>
  add_recipe(knn_recipe) |>
  add_model(knn_spec)

# Train the model using the training data
knn_fit <- fit(knn_workflow, data = training_data)

# Predict tournament results using the testing data
knn_predictions <- predict(knn_fit, testing_data) |>
  bind_cols(testing_data)

# Calculate the accuracy
knn_accuracy <- knn_predictions |>
  metrics(truth = tourney_finish, estimate = .pred_class) |>
  filter(.metric == "accuracy")

knn_accuracy

# Create the confusion matrix
knn_confusion <- knn_predictions |>
  conf_mat(truth = tourney_finish, estimate = .pred_class)

knn_confusion
```
